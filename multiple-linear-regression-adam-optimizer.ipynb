# Multiple Linear Regression with Adam Optimizer

This project uses Adam instead of the basic gradient descent update to train a multiple linear regression model. The dataset contains fake values for practice.

```python
import numpy as np
class Adam:
  def __init__(self,params,alpha=0.1,beta1=0.9,beta2=0.999,ep=1e-8):
    self.params = params
    self.alpha =alpha
    self.beta1 = beta1
    self.beta2 = beta2
    self.ep = ep
    self.t = 0
    self.m = []
    self.v = []
    # 1. Create first- and second-moment arrays for every parameter.
    for p in self.params:
      self.m.append(np.zeros_like(p,dtype=float))
      self.v.append(np.zeros_like(p,dtype=float))
  def Upd_moments(self,grads):
    self.t += 1
    # 2. Update the moving averages of gradients and squared gradients.
    for i in range(len(self.params)):
      self.m[i] = self.beta1*self.m[i]+(1-self.beta1)*grads[i]
      self.v[i] = self.beta2*self.v[i]+(1-self.beta2)*(grads[i]**2)
  def corrected_version(self):
    m_hat = []
    v_hat = []
    # 3. Correct the initial bias in the moment estimates.
    for i in range(len(self.params)):
      m_hat.append(
          self.m[i]/(1-self.beta1**self.t)
      )
      v_hat.append(
          self.v[i]/(1-self.beta2**self.t)
      )
    return m_hat,v_hat
  def step(self,grads):
    self.Upd_moments(grads)
    m_hat,v_hat = self.corrected_version()
    # 4. Update each parameter using the corrected moments.
    for i in range(len(self.params)):
      self.params[i] -= (
          self.alpha*m_hat[i]/(
              np.sqrt(v_hat[i])+self.ep
          )
      )
```

```python
import numpy as np

# Fake numbers, just for training

# 5. Create the input features and target values.
X = np.array([
    [1.0, 2.0],
    [2.0, 3.0],
    [3.0, 4.0],
    [4.0, 5.0]
])

y = np.array([5.0, 8.0, 11.0, 14.0])

# 6. Initialize one weight per feature and a scalar bias.
w = np.zeros(X.shape[1])
b = np.array(0.0)

# 7. Calculate the squared-error cost.
def cost(X,y,w,b):
  m = len(y)
  error = X@w+b-y
  cost = np.sum(error**2)/(2*m)
  return cost

# 8. Calculate the gradients for the weights and bias.
def gradient(X,y,w,b):
  m=len(y)
  error = X@w+b-y
  Gw = (X.T@error)/m
  Gb = np.sum(error)/m
  return Gw,Gb

# 9. Run Adam and update the parameters for each iteration.
def CompAdam(X,y,w,b,num_iters):
  optimizer = Adam([w,b])
  for i in range(num_iters):
    Gw,Gb = gradient(X,y,w,b)
    optimizer.step([Gw,np.array(Gb)])
  return w,b

# 10. Normalize the input features, train with Adam, and print predictions.
mu = np.mean(X, axis=0)
sig = np.std(X, axis=0)

x_s = (X - mu) / sig
w1,b1 = CompAdam(x_s,y,w,b,num_iters=1000)

# With scaling
print(x_s@w1+b)
```
