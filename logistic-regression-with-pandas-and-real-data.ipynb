```python
import pandas as pdimport numpy as np# 1. Load the real dataset and give each column a name.url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"columns = [    "variance",    "skewness",    "curtosis",    "entropy",    "class"]df = pd.read_csv(    url,    header=None,    names=columns)# 2. X contains the four input features; y contains the class labels.X = df[    ["variance", "skewness", "curtosis", "entropy"]].to_numpy(dtype=float)y = df["class"].to_numpy(dtype=float)# 3. Start with one weight for each feature and one bias.w = np.zeros(X.shape[1])b=0.0def costF(X,y,w,b):  m=len(y)  z = X@w+b  # 4. The sigmoid turns each linear score into a probability between 0 and 1.  f = 1/(1+np.exp(-z))  # 5. Binary cross-entropy measures how well the predicted probabilities match the labels.  cost = -np.sum(y*np.log(f)+(1-y)*np.log(1-f))/m  return costdef Gradient(X,y,w,b,alpha,num_iters):  m=len(y)  z = X@w+b  for i in range(num_iters):    z = X@w+b    f = 1/(1+np.exp(-z))    # 6. These are the partial derivatives of the cost with respect to w and b.    dj_dw = X.T@(f-y)/m    dj_db = np.sum(f-y)/m    # 7. Update the weights and bias by moving opposite the gradient.    w = w-alpha*dj_dw    b = b-alpha*dj_db  return w,bnew_w,new_b = Gradient(X,y,w,b,alpha=0.05,num_iters=500)print(costF(X,y,new_w,new_b))
```
