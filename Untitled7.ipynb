```python
#Hello, this is a simple project for gradient descent(multiple features) with scaling and without, we will start without the scaling so i will try explaining what i did and why
#also im ranking every thing i need to explain later in README to make it simple
#before saying anything, this project have fake numbers , just for training
import numpy as np

X = np.array([#creating an array that have 2 features, the first column is the first feature, and the sec column is the sec feature,
    [1.0, 2.0],
    [2.0, 3.0],
    [3.0, 4.0],
    [4.0, 5.0]
])
y = np.array([5.0, 8.0, 11.0, 14.0])#the target or the output

w = np.zeros(X.shape[1])#1: in gradient descent we usually make w and b = zeros, so we made w here have X.shape[1](number of columns) but of zeros, here X have 2 columns, so w = [0,0] or [0.0,0.0] i will explain why in README file,
b = 0.0#same thing but b is always one number so 0.0 is enough
def CompCost(X,y,w,b):#here its stands for Computing the cost, and its a tool used to measure good is a model,better if closer to the 0 ,
  m=len(y)#y have 4 cells so m=4
  error = (X@w+b)-y#so X@w+b is the prediction of the target, and then subtract y(the real target) from it to compare
  cost = np.sum(error**2)/(2*m)#2: just the normal cost func, explained in README
  return cost#we need to return it so when we call out the func we can use cost
def CompGradient(X,y,w,b):#Computing the Gradient: its simillar to CompCost so i will skip every thing i explained
  m=len(y)
  error = X@w+b-y
  Gw = (X.T@error)/m#3: so that is the derivative function with respect to w, explained in README
  Gb = np.sum(error)/m#same as 3 but with respect to b
  return Gw,Gb
def GradientDescent(X,y,w,b,alpha,n):#now we will use alpha and n or the number of iter
  for i in range(n):#using the loop of range of n
    Gw,Gb=CompGradient(X,y,w,b)#calling out the func to use it
    w = w-alpha*Gw#updating the w with the gradient descent formula
    b= b-alpha*Gb#same thing but with b
  return w,b
F_w,F_b = GradientDescent(X,y,w,b,alpha=0.01,n=1000)#now we will use the gradient func , F_w stands for Final w and F_b also
print("without scalling output",X@F_w+F_b)#now multiplying X with the new updated w and b
print("without scalling cost or loss:",CompCost(X,y,F_w,F_b))#checking the cost func
mu=np.mean(X,axis=0)#4: this is the Z-score normalization, explained in README
sig = np.std(X,axis=0)#same as 4
x_s = (X-mu)/sig#same as 4
F_w1,F_b1 = GradientDescent(x_s,y,w,b,alpha=0.1,n=1000)#calc the new values and putting the outputs in F_w1 and F_b1
print("with scalling",x_s@F_w1+F_b1)#now seeing scaled version
print("without scalling cost or loss:",CompCost(x_s,y,F_w1,F_b1))#same
```
