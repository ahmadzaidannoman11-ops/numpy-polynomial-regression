```python
import numpy as np
num_ores = np.array([64, 8_000, 1_000_000, 125_000_000])

lapis = np.array([53_153, 6_679_871, 831_652_567, 103_996_242_510])
X = np.stack((np.ones_like(num_ores),num_ores),axis=1)
rng = np.random.default_rng(42)
indices = rng.permutation(len(X))
split = int(0.75*len(indices))
train = indices[:split]
test = indices[split:]
X_train = X[train]
y_train = lapis[train]
X_test = X[test]
y_test = lapis[test]
theta_train = np.linalg.lstsq(X_train,y_train,rcond=None)[0]
a = float(input("Number of ores: "))
X_new = np.array([1., a])
print(X_new @ theta_train)
```
